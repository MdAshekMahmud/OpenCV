Drawing Line

```python
import cv2
import matplotlib.pyplot as plt

# Read image
image = cv2.imread("/content/drive/MyDrive/Colab Notebooks/datasets/opencv-python-free-course-code/03_Image_Annotation/Apollo_11_Launch.jpg")

# Create a copy
line_image = image.copy()

# Line properties
pt1 = (150, 450)
pt2 = (1200, 450)
color = (255, 0, 0)   # Blue in BGR
thickness = 3

# Draw line on the copy
cv2.line(line_image, pt1, pt2, color, thickness)

# Display Original vs Line Image
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(cv2.cvtColor(line_image, cv2.COLOR_BGR2RGB))
plt.title("Image with Line")
plt.axis("off")

plt.tight_layout()
plt.show()
```

Drawing Rectangle

```python
import cv2
import matplotlib.pyplot as plt

# Read image
image = cv2.imread("/content/drive/MyDrive/Colab Notebooks/datasets/opencv-python-free-course-code/04_Image_Enhancement/building-windows.jpg")

# Create a copy
rectangle_image = image.copy()

# Rectangle properties
pt1 = (50, 50)            # Top-left corner (x, y)
pt2 = (700, 500)          # Bottom-right corner (x, y)
color = (0, 255, 0)       # Green (BGR)
thickness = 5             # Border thickness(-1 for filled rectangle)

# Draw rectangle
cv2.rectangle(rectangle_image, pt1, pt2, color, thickness)

# Display images
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
plt.title("Original Image")
# plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(cv2.cvtColor(rectangle_image, cv2.COLOR_BGR2RGB))
plt.title("Rectangle")
plt.axis("off")

plt.tight_layout()
plt.show()
```

Drawing Circle

```python
import cv2
import matplotlib.pyplot as plt

# Read image
image = cv2.imread("/content/drive/MyDrive/Colab Notebooks/datasets/opencv-python-free-course-code/01_Getting_Started_with_Images/coca-cola-logo.png")

# Create a copy
circle_image = image.copy()

# Circle properties
center = (490, 290)      # (x, y)
radius = 170             # Radius in pixels
color = (0, 0, 199)      # Red (BGR)
thickness = cv2.FILLED   # Border thickness

# Draw circle
cv2.circle(circle_image, center, radius, color, thickness)

# Display Original vs Circle
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(cv2.cvtColor(circle_image, cv2.COLOR_BGR2RGB))
plt.title("Circle")
plt.axis("off")

plt.tight_layout()
plt.show()
```

Adding Text

```python
import cv2
import matplotlib.pyplot as plt

# Read image
image = cv2.imread("/content/drive/MyDrive/Colab Notebooks/datasets/opencv-python-free-course-code/04_Image_Enhancement/building-windows.jpg")

# Create a copy
text_image = image.copy()

# Text properties
text = "Hello World!"
org = (100, 100)                  # Bottom-left corner of text (x, y)
font = cv2.FONT_HERSHEY_SIMPLEX   # Font
fontScale = 2
color = (0, 255, 255)             # Yellow (BGR)
thickness = 3
lineType = cv2.LINE_AA

# Put text
cv2.putText(text_image, text, org, font, fontScale, color, thickness, lineType)

# Display
plt.imshow(cv2.cvtColor(text_image, cv2.COLOR_BGR2RGB))
plt.axis("off")
plt.show()
```
