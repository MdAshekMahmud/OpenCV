Resizing and Scaling Images - `cv2.resize()`

```python
import cv2
import matplotlib.pyplot as plt

# Resize image
resize = cv2.resize(image, (400, 400))

# Display Original vs Resized
plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
plt.title("Original")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(cv2.cvtColor(resize, cv2.COLOR_BGR2RGB))
plt.title("Resized (400 × 400)")
plt.axis("off")

plt.show()
```

Cropping Images- using Slicing

```python
image.shape
```

```python
cropped_image = image[400:500, :100]
plt.imshow(cv2.cvtColor(cropped_image, cv2.COLOR_BGR2RGB))
plt.title("Cropped Image")
plt.axis("off")
plt.show()
```

Image Rotation and Flipping

```python
width = image.shape[1]
height = image.shape[0]

# image.shape[1] -> width (x-coordinate)
# image.shape[0] -> height (y-coordinate)

# Center of image
center = (width / 2, height / 2)

# Rotation matrix
M = cv2.getRotationMatrix2D(center=center, angle=90, scale=1)

# Rotate image
rotated_image = cv2.warpAffine(src=image, M=M, dsize=(width, height))

# Display Original vs Rotated
plt.figure(figsize=(10, 5))

# Original image
plt.subplot(1, 2, 1)
plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
plt.title("Original Image")
plt.axis("off")

# Rotated image
plt.subplot(1, 2, 2)
plt.imshow(cv2.cvtColor(rotated_image, cv2.COLOR_BGR2RGB))
plt.title("Rotated Image (90°)")
plt.axis("off")

plt.show()
```

```python
import matplotlib.pyplot as plt
import cv2

image = cv2.imread("/content/drive/MyDrive/Colab Notebooks/datasets/opencv-python-free-course-code/02_Basic_Image_Manipulation/New_Zealand_Lake.jpg")

# Flip images
horizontal_flip = cv2.flip(src=image, flipCode=1)
vertical_flip = cv2.flip(src=image, flipCode=0) # Upside-down effect.
both_flip = cv2.flip(src=image, flipCode=-1) # Equivalent to a 180-degree

# Display images
plt.figure(figsize=(12, 8))

# Original
plt.subplot(2, 2, 1)
plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
plt.title("Original Image")
plt.axis("off")

# Horizontal Flip
plt.subplot(2, 2, 2)
plt.imshow(cv2.cvtColor(horizontal_flip, cv2.COLOR_BGR2RGB))
plt.title("Horizontal Flip (flipCode=1)")
plt.axis("off")

# Vertical Flip
plt.subplot(2, 2, 3)
plt.imshow(cv2.cvtColor(vertical_flip, cv2.COLOR_BGR2RGB))
plt.title("Vertical Flip (flipCode=0)")
plt.axis("off")

# Both Flip
plt.subplot(2, 2, 4)
plt.imshow(cv2.cvtColor(both_flip, cv2.COLOR_BGR2RGB))
plt.title("Both Flip (flipCode=-1)")
plt.axis("off")

plt.tight_layout()
plt.show()
```
