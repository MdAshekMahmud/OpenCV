### Choose Action for the User-Provided Image
Select 'show' to display the image or 'save' to save it. If saving, provide a filename.

```python
action_choice = "show" # @param ["show", "save"]
output_filename = "my_output_image.png" # @param {type:"string"}
```

```python
if user_image is not None:
    if action_choice == "show":
        cv2_imshow(user_image)
    elif action_choice == "save":
        try:
            cv2.imwrite(output_filename, user_image)
            print(f"Image saved successfully as {output_filename}")
        except Exception as e:
            print(f"Error saving image: {e}")
    else:
        print("Invalid action chosen. Please select 'show' or 'save'.")
else:
    print("No image loaded to perform the action.")
```
